### **Cell and Morphology Selection**

**<span style="color: darkred;">This is notebook 1/x of the gene expression prediction of cortical motor neuron morphology analysis.</span>**

- Initial files are sourced from the berenslab/mini-atlas GitHub repository: https://github.com/berenslab/mini-atlas

    They are additionally available within the _../data/_ folder in this repository.

- The following script includes interrogation of the input files and identifying the cells and morphologies to be excluded in the downstream analyses.

- All output data files are available in:

    _../data/processed/_

```python
# Libraries
import os
import pandas as pd
```

**There are 1329 neurons in the counts data that passed quality control and should be present in the Gene Counts and Metadata tables.**

**According to the paper, the following criteria is considered low quality and cause for exclusion:**
Cells with the highest correlation below 0.4 (78 cells); cells that would be assigned to non-neural t-types, presumably owing to RNA contamination (14 cells); cells with the highest correlation less than 0.02 above the maximal correlation in one of the other two large transcriptomic groups (5 cells).

**<span style="color: darkred;">1232 neurons remain after this quality control.</span>**

**The paper then did a t-type assignment that returned t-type confidence scores. The following resulted in cell exclusion:**
Cells assigned to an excitatory t-type, despite having inhibitory firing, morphology, and/or soma depth location (such as L1). The most likely cause of this was RNA contamination from excitatory cells, which are much more abundant than inhibitory cells in the mouse cortex. This was the case for 4 cells, and they were excluded. One cell was also probably located outside MOp, based on the slice anatomy, and was excluded as well. 

**<span style="color: darkred;">1227 neurons remained after these rounds of t-typing.</span>** This is the final list of cell types that will be used for this analysis, as we are not looking at t-type necessarily, but the assignment of t-type identified RNA contamination and other valid reasons for sample exclusion.

The berenslab/mini-atlas GitHub also describes an incorrect morphological reconstruction for a sample (**20171207sample1**). This morphology will therefore be excluded in this downstream analysis.

### <div style = 'background-color:PapayaWhip'> **Load Gene Counts and Morphological Data**</div>

```python
# Gene Counts paths
exon_counts = pd.read_csv("../data/m1_patchseq_exon_counts.csv.gz", sep=",", compression='gzip', index_col=0)
intron_counts = pd.read_csv("../data/m1_patchseq_intron_counts.csv.gz", sep=",", compression='gzip', index_col=0)
```

```python
# Morphological features
morph_features = pd.read_csv("../data/m1_patchseq_morph_features.csv", sep=",", index_col=0)
morph_zprofiles = pd.read_csv("../data/m1_patchseq_morph_zprofiles.csv", sep=",", index_col=0)
```

### <div style = 'background-color:PapayaWhip'> **Sample Exclusion**</div>

```python
# Print initial table sizes
print("exon_counts:", exon_counts.shape)
print("intron_counts:", intron_counts.shape)
print("morph_features:", morph_features.shape)
print("morph_zprofiles:", morph_zprofiles.shape)
```

```python
# Remove gene counts and morphology data of selected cells

# Remove 20171207sample1 from both morphology .csv files
morph_features = morph_features[morph_features.index != "20171207_sample_1"]
morph_zprofiles = morph_zprofiles[morph_zprofiles.index != "20171207_sample_1"]

# Remove excluded neurons from datasets
to_remove = [
    "20180315_sample_6",
    "20180321_sample_4",
    "20181116_sample_10",
    "20190325_sample_2",
    "20190611_sample_1",
    "20190611_sample_11",
    "20190704_sample_2",
]

# Morphology tables
morph_features  = morph_features[~morph_features.index.isin(to_remove)]
morph_zprofiles = morph_zprofiles[~morph_zprofiles.index.isin(to_remove)]
# Gene count tables
exon_counts   = exon_counts.drop(columns=to_remove, errors="ignore")
intron_counts = intron_counts.drop(columns=to_remove, errors="ignore")
```

```python
# Check table sizes
print("exon_counts:", exon_counts.shape)
print("intron_counts:", intron_counts.shape)
print("morph_features:", morph_features.shape)
print("morph_zprofiles:", morph_zprofiles.shape)
```

```python
# Save files as new processed data for downstream use
meta_data.to_csv("../data/processed/m1_patchseq_meta_data_clean.csv", sep="\t")
morph_features.to_csv("../data/processed/m1_patchseq_morph_features_clean.csv", sep=",")
morph_zprofiles.to_csv("../data/processed/m1_patchseq_morph_zprofiles_clean.csv", sep=",")
exon_counts.to_csv("../data/processed/m1_patchseq_exon_counts_clean.csv.gz", sep=",", compression="gzip")
intron_counts.to_csv("../data/processed/m1_patchseq_intron_counts_clean.csv.gz", sep=",", compression="gzip")
```
