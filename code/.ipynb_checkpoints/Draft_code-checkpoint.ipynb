## Cell Selection

**There are 1329 neurons in the counts data that passed quality control and are present in the Gene Counts and Metadata tables.**

**According to the paper, the following criteria is considered low quality and cause for exclusion:**
Cells with the highest correlation below 0.4 (78 cells); cells that would be assigned to non-neural t-types, presumably owing to RNA contamination (14 cells); cells with the highest correlation less than 0.02 above the maximal correlation in one of the other two large transcriptomic groups(5 cells).

**1232 neurons remain after this quality control.**

**The paper then did a t-type assignment that returned t-type confidence scores. The following resulted in cell exclusion:**
Cells assigned to an excitatory t-type, despite having inhibitory firing, morphology, and/or soma depth location (such as L1). The most likely cause of this was RNA contamination from excitatory cells, which are much more abundant than inhibitory cells in the mouse cortex. This was the case for 4 cells, and they were excluded. One cell was also probably located outside MOp, based on the slice anatomy, and was excluded as well. 

**1227 neurons remained after these rounds of t-typing.** This is the final list of cell types that will be used for this analysis, as we are not looking at t-type necessarily, but the assignment of t-type identified RNA contamination and other valid reasons for sample exclusion.

The berenslab/mini-atlas GitHub also describes an incorrect morphological reconstruction for a sample (**20171207sample1**). This morphology will therefore be excluded in this downstream analysis.

## Define Morphologies

The **morph_zprofiles** is the normalised z profile of each neuron. 
Represents the neuron's normalised depth profile perpendicular to the pia. Binned into 20 bins and treated as a set of 20 features for downstream analyses. 

The original paper defined a set of 75 features initially. They then excluded features with coefficient of variation below 0.25 (among

### _Define morphologies using several features via PCA_

I'm in Python. 

What I have is:
1322 neurons with gene expression data (exon counts and intron counts)
645 of which have morphological features (371 inhibitory, 274 excitatory)

There are 35 defined morphological excitatory neuron features, and 50 inhibitory neuron morphological features (all of which are numerical)

What the concept is:
- I need to run a PCA on the expression matrix to get Expression PCs 
- I need to run a PCA on the inhibitory neurons to get morphology PCs
- I need to run a PCA on the excitatory neurons to get morphology PCs
Once I have these, I can correlate the Expression PCs with Morphology PCs to identify overlapping biological signals. This can either be done using pairwise PC correlations or canonical correlation analysis.

Do I need to split my gene expression into inhibitory and excitatory neurons and run separate PC analysis for those? 
My gene count matrix currently has 42k genes. Should I first narrow it down to the top 2000 most variable genes? Should I only be using the neurons that have associated morphological features? What statistical validation can I do on the results I get?

```python
# Split datasets into excitatory and inhibitory neural cells

# Combine Exon and Intron Counts
# Matrices are structured as: rows = genes, columns = samples
combined_counts = exon_counts + intron_counts
```

```python
combined_counts
```

```python
# Identify Samples for Each Class from 'morph_features'
exc_cells = morph_features[morph_features["cell class"] == "exc"].index
inh_cells = morph_features[morph_features["cell class"] == "inh"].index
```

```python
# Filter counts and transpose so rows=cells and columns=genes
exc_matrix_ready = combined_counts[
    combined_counts.columns.intersection(exc_cells)
].T
inh_matrix_ready = combined_counts[
    combined_counts.columns.intersection(inh_cells)
].T
```

```python
inh_matrix_ready
```

```python
# Run function from berenslab/mini-atlas/code/ 
%matplotlib inline
from rnaseqTools import geneSelection

exc_selected_mask = geneSelection(
    data=exc_matrix_ready.values,
    threshold=32,
    atleast=10,
    yoffset=0.002,
    decay=1.5,
    n=2000,
    plot=True,
)
```

```python
# Extract the final names of the 2,000 highly variable excitatory genes
exc_vargenes = exc_matrix_ready.columns[exc_selected_mask]
exc_final_df = exc_matrix_ready[exc_vargenes]
```

```python
exc_final_df
```

```python
inh_selected_mask = geneSelection(
    data=inh_matrix_ready.values,
    threshold=32,
    atleast=10,
    yoffset=0.002,
    decay=1.5,
    n=2000,
    plot=True,
)
```

```python
inh_vargenes = inh_matrix_ready.columns[inh_selected_mask]
inh_final_df = inh_matrix_ready[inh_vargenes]
```

```python
inh_final_df
```

```python
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# --- 1. CPM Normalisation ---
# Divide each cell (row) by its total count across the 2,000 genes, then multiply by 1M
exc_sums = exc_final_df.sum(axis=1)
exc_cpm = exc_final_df.div(exc_sums, axis=0) * 1e6

# --- 2. Log2(x + 1) Transformation ---
exc_log = np.log2(exc_cpm + 1)

# --- 3. Z-score Scaling ---
scaler_exc = StandardScaler()
exc_scaled_data = scaler_exc.fit_transform(exc_log)
exc_scaled_df = pd.DataFrame(
    exc_scaled_data, index=exc_log.index, columns=exc_log.columns
)

# --- 4. PCA Calculation (Top 10 PCs) ---
pca_exc = PCA(n_components=10)
exc_pc_coords = pca_exc.fit_transform(exc_scaled_df)

# Store in a clean DataFrame
exc_pc_cols = [f"Expression_PC{i+1}" for i in range(10)]
exc_expression_pcs = pd.DataFrame(
    exc_pc_coords, index=exc_scaled_df.index, columns=exc_pc_cols
)

# --- 5. Plotting ---
plt.figure(figsize=(5.5, 4.5))
plt.scatter(
    exc_expression_pcs["Expression_PC1"],
    exc_expression_pcs["Expression_PC2"],
    s=15,
    alpha=0.7,
    c="#1f77b4",
)
exc_var = pca_exc.explained_variance_ratio_ * 100
plt.title(f"Expression PCA: Excitatory Neurons (N={len(exc_scaled_df)})")
plt.xlabel(f"PC1 ({exc_var[0]:.1f}% variance)")
plt.ylabel(f"PC2 ({exc_var[1]:.1f}% variance)")
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()

print(
    f"Total variance explained by top 10 Excitatory PCs: {np.sum(exc_var):.1f}%"
)
```

```python
# --- 1. CPM Normalisation ---
inh_sums = inh_final_df.sum(axis=1)
inh_cpm = inh_final_df.div(inh_sums, axis=0) * 1e6

# --- 2. Log2(x + 1) Transformation ---
inh_log = np.log2(inh_cpm + 1)

# --- 3. Z-score Scaling ---
scaler_inh = StandardScaler()
inh_scaled_data = scaler_inh.fit_transform(inh_log)
inh_scaled_df = pd.DataFrame(
    inh_scaled_data, index=inh_log.index, columns=inh_log.columns
)

# --- 4. PCA Calculation (Top 10 PCs) ---
pca_inh = PCA(n_components=10)
inh_pc_coords = pca_inh.fit_transform(inh_scaled_df)

# Store in a clean DataFrame
inh_pc_cols = [f"Expression_PC{i+1}" for i in range(10)]
inh_expression_pcs = pd.DataFrame(
    inh_pc_coords, index=inh_scaled_df.index, columns=inh_pc_cols
)

# --- 5. Plotting ---
plt.figure(figsize=(5.5, 4.5))
plt.scatter(
    inh_expression_pcs["Expression_PC1"],
    inh_expression_pcs["Expression_PC2"],
    s=15,
    alpha=0.7,
    c="#ff7f0e",
)
inh_var = pca_inh.explained_variance_ratio_ * 100
plt.title(f"Expression PCA: Inhibitory Neurons (N={len(inh_scaled_df)})")
plt.xlabel(f"PC1 ({inh_var[0]:.1f}% variance)")
plt.ylabel(f"PC2 ({inh_var[1]:.1f}% variance)")
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()

print(
    f"Total variance explained by top 10 Inhibitory PCs: {np.sum(inh_var):.1f}%"
)
```

Since your goal is to find overlapping signals, implement Pairwise PC Correlation first, followed by CCA. They tell you different things:
- Pairwise PC Correlation (scipy.stats.pearsonr): Tells you if the dominant independent axes of variance line up naturally. (e.g., "Does Expression PC₁ correlate with Morphology PC₂?")
- CCA (sklearn.cross_decomposition.CCA): Finds a novel rotation of your features specifically engineered to maximize the correlation between expression and morphology. It forces a relationship. This is powerful but highly prone to overfitting with small sample sizes (N < 400).

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import pearsonr
from statsmodels.stats.multitest import multipletests

# --- Step 1: Ensure row sorting matches exactly ---
# Since cells are identical, sort both dataframes by their index to align row orders
exc_expression_pcs = exc_expression_pcs.sort_index()
exc_morphology_pcs = exc_morphology_pcs.sort_index()

X_exc_expr = exc_expression_pcs.values  # N x 10
Y_exc_morph = exc_morphology_pcs.values  # N x 10
n_cells_exc = X_exc_expr.shape[0]

# --- Step 2: Compute True Pearson r and Analytical p-values ---
true_r_matrix_exc = np.zeros((10, 10))
analytical_p_matrix_exc = np.zeros((10, 10))

for i in range(10):
    for j in range(10):
        r, p = pearsonr(X_exc_expr[:, i], Y_exc_morph[:, j])
        true_r_matrix_exc[i, j] = r
        analytical_p_matrix_exc[i, j] = p

# --- Step 3: Permutation Testing (1,000 iterations) ---
n_permutations = 1000
greater_than_null_counts_exc = np.zeros((10, 10))

np.random.seed(42)  # For reproducibility

for p_iter in range(n_permutations):
    # Shuffle only the morphology row indices to break the biological link
    shuffled_indices = np.random.permutation(n_cells_exc)
    Y_morph_shuffled = Y_exc_morph[shuffled_indices, :]

    for i in range(10):
        for j in range(10):
            r_null, _ = pearsonr(X_exc_expr[:, i], Y_morph_shuffled[:, j])
            # Two-tailed absolute magnitude comparison
            if np.abs(r_null) >= np.abs(true_r_matrix_exc[i, j]):
                greater_than_null_counts_exc[i, j] += 1

empirical_p_matrix_exc = greater_than_null_counts_exc / n_permutations

# --- Step 4: Multiple Testing Correction (FDR) ---
flat_r_exc = true_r_matrix_exc.flatten()
flat_p_empirical_exc = empirical_p_matrix_exc.flatten()

# Apply Benjamini-Hochberg FDR correction on permutation p-values
_, fdr_p_empirical_exc, _, _ = multipletests(
    flat_p_empirical_exc, alpha=0.05, method="fdr_bh"
)
fdr_p_empirical_exc = fdr_p_empirical_exc.reshape(10, 10)

# --- Step 5: Plotting the Results ---
plt.figure(figsize=(8, 6.5))
significance_mask = fdr_p_empirical_exc < 0.05

annot_labels = np.empty((10, 10), dtype=object)
for i in range(10):
    for j in range(10):
        star = "*" if significance_mask[i, j] else ""
        annot_labels[i, j] = f"{true_r_matrix_exc[i, j]:.2f}{star}"

sns.heatmap(
    true_r_matrix_exc,
    annot=annot_labels,
    fmt="",
    cmap="RdBu_r",
    vmin=-1,
    vmax=1,
    xticklabels=[f"Morph_PC{i+1}" for i in range(10)],
    yticklabels=[f"Expr_PC{i+1}" for i in range(10)],
)
plt.title(
    "Excitatory Neurons: PC Correlations\n(* indicates Permutation FDR p < 0.05)"
)
plt.tight_layout()
plt.show()
```

```python
# --- 1. Extract Gene Loadings for Expr_PC4 ---
# pca_exc.components_ has shape (n_components, n_features)
# Index 3 corresponds to PC4
expr_pc4_loadings = pd.Series(pca_exc.components_[3], index=exc_scaled_df.columns)

print("=== TOP GENES DRIVING EXPR_PC4 ===")
print("Positive drivers (associated with high Morph_PC1):")
print(expr_pc4_loadings.sort_values(ascending=False).head(10))
print("\nNegative drivers:")
print(expr_pc4_loadings.sort_values(ascending=True).head(10))

print("\n" + "="*40 + "\n")

# --- 2. Extract Morphological Feature Loadings for Morph_PC1 ---
# pca_morph_exc.components_ has shape (n_components, n_features)
# Index 0 corresponds to PC1
morph_pc1_loadings = pd.Series(pca_morph_exc.components_[0], index=exc_morph_scaled_df.columns)

print("=== TOP MORPHOLOGY FEATURES DRIVING MORPH_PC1 ===")
print("Positive structural features:")
print(morph_pc1_loadings.sort_values(ascending=False).head(5))
print("\nNegative structural features:")
print(morph_pc1_loadings.sort_values(ascending=True).head(5))
```

```python
# --- Step 1: Ensure row sorting matches exactly ---
inh_expression_pcs = inh_expression_pcs.sort_index()
inh_morphology_pcs = inh_morphology_pcs.sort_index()

X_inh_expr = inh_expression_pcs.values  # M x 10
Y_inh_morph = inh_morphology_pcs.values  # M x 10
n_cells_inh = X_inh_expr.shape[0]

# --- Step 2: Compute True Pearson r ---
true_r_matrix_inh = np.zeros((10, 10))
analytical_p_matrix_inh = np.zeros((10, 10))

for i in range(10):
    for j in range(10):
        r, p = pearsonr(X_inh_expr[:, i], Y_inh_morph[:, j])
        true_r_matrix_inh[i, j] = r
        analytical_p_matrix_inh[i, j] = p

# --- Step 3: Permutation Testing (1,000 iterations) ---
greater_than_null_counts_inh = np.zeros((10, 10))

for p_iter in range(n_permutations):
    shuffled_indices = np.random.permutation(n_cells_inh)
    Y_morph_shuffled = Y_inh_morph[shuffled_indices, :]

    for i in range(10):
        for j in range(10):
            r_null, _ = pearsonr(X_inh_expr[:, i], Y_morph_shuffled[:, j])
            if np.abs(r_null) >= np.abs(true_r_matrix_inh[i, j]):
                greater_than_null_counts_inh[i, j] += 1

empirical_p_matrix_inh = greater_than_null_counts_inh / n_permutations

# --- Step 4: Multiple Testing Correction (FDR) ---
flat_r_inh = true_r_matrix_inh.flatten()
flat_p_empirical_inh = empirical_p_matrix_inh.flatten()

_, fdr_p_empirical_inh, _, _ = multipletests(
    flat_p_empirical_inh, alpha=0.05, method="fdr_bh"
)
fdr_p_empirical_inh = fdr_p_empirical_inh.reshape(10, 10)

# --- Step 5: Plotting the Results ---
plt.figure(figsize=(8, 6.5))
significance_mask_inh = fdr_p_empirical_inh < 0.05

annot_labels_inh = np.empty((10, 10), dtype=object)
for i in range(10):
    for j in range(10):
        star = "*" if significance_mask_inh[i, j] else ""
        annot_labels_inh[i, j] = f"{true_r_matrix_inh[i, j]:.2f}{star}"

sns.heatmap(
    true_r_matrix_inh,
    annot=annot_labels_inh,
    fmt="",
    cmap="RdBu_r",
    vmin=-1,
    vmax=1,
    xticklabels=[f"Morph_PC{i+1}" for i in range(10)],
    yticklabels=[f"Expr_PC{i+1}" for i in range(10)],
)
plt.title(
    "Inhibitory Neurons: PC Correlations\n(* indicates Permutation FDR p < 0.05)"
)
plt.tight_layout()
plt.show()
```
